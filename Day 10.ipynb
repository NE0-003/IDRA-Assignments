```python
import numpy as np
import pandas as pd

from google.colab import files

uploaded = files.upload()

df = pd.read_csv("Day10_Flight_Operations_Dataset.csv")

df["Flight_Date"] = pd.to_datetime(df["Flight_Date"], errors="coerce")

print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
df.head()
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

display(df.head())
display(df.tail())
```

```python
df.info()
```

```python
missing_values = df.isna().sum().sort_values(ascending=False)
display(missing_values.to_frame("Missing_Count"))

print("Duplicate rows:", df.duplicated().sum())

unique_counts = df.nunique().sort_values()
display(unique_counts.to_frame("Unique_Values"))
```

```python
display(df.describe().round(2))

display(df.describe(include="object").T)
```

```python
selected_columns = df[
    ["Flight_ID", "Airline", "Origin", "Destination",
     "Passengers", "Delay_Minutes", "Flight_Status",
     "Passenger_Satisfaction"]
]

display(selected_columns.head(10))

display(
    df.loc[:, ["Flight_ID", "Airline", "Average_Ticket_Price", "Passenger_Satisfaction"]].head(10)
)
```

```python
delayed_30 = df[df["Delay_Minutes"] > 30]
print("Flights delayed more than 30 minutes:", len(delayed_30))
display(delayed_30[
    ["Flight_ID", "Airline", "Origin", "Destination", "Delay_Minutes", "Weather"]
].head(10))

high_satisfaction = df[df["Passenger_Satisfaction"] >= 8]
print("Flights with satisfaction >= 8:", len(high_satisfaction))

high_delay_low_satisfaction = df[
    (df["Delay_Minutes"] > 30) &
    (df["Passenger_Satisfaction"] < 7)
]
print("Flights with delay > 30 min and satisfaction < 7:", len(high_delay_low_satisfaction))
display(high_delay_low_satisfaction.head(10))
```

```python
display(
    df.sort_values("Delay_Minutes", ascending=False)[
        ["Flight_ID", "Airline", "Origin", "Destination", "Delay_Minutes", "Flight_Status"]
    ].head(10)
)

display(
    df.sort_values("Passenger_Satisfaction", ascending=False)[
        ["Flight_ID", "Airline", "Passenger_Satisfaction", "Delay_Minutes"]
    ].head(10)
)
```

```python
df["Load_Factor_%"] = (df["Passengers"] / df["Seat_Capacity"] * 100).round(2)

df["Revenue_Estimate"] = df["Passengers"] * df["Average_Ticket_Price"]

df["Delay_Category"] = pd.cut(
    df["Delay_Minutes"],
    bins=[-np.inf, 0, 15, 60, np.inf],
    labels=["On time", "Minor (1–15 min)", "Moderate (16–60 min)", "Severe (>60 min)"]
)

display(
    df[
        ["Flight_ID", "Passengers", "Seat_Capacity", "Load_Factor_%",
         "Average_Ticket_Price", "Revenue_Estimate", "Delay_Minutes", "Delay_Category"]
    ].head(10)
)
```

```python
airline_summary = (
    df.groupby("Airline")
      .agg(
          Flights=("Flight_ID", "count"),
          Passengers=("Passengers", "sum"),
          Avg_Delay_Min=("Delay_Minutes", "mean"),
          Avg_Satisfaction=("Passenger_Satisfaction", "mean"),
          Avg_Load_Factor=("Load_Factor_%", "mean"),
          Estimated_Revenue=("Revenue_Estimate", "sum")
      )
      .sort_values("Estimated_Revenue", ascending=False)
      .round(2)
)

display(airline_summary)
```

```python
status_summary = (
    df.groupby("Flight_Status")
      .agg(
          Flights=("Flight_ID", "count"),
          Avg_Delay_Min=("Delay_Minutes", "mean"),
          Avg_Satisfaction=("Passenger_Satisfaction", "mean")
      )
      .sort_values("Flights", ascending=False)
      .round(2)
)

display(status_summary)
```

```python
class_summary = (
    df.groupby("Travel_Class")
      .agg(
          Flights=("Flight_ID", "count"),
          Passengers=("Passengers", "sum"),
          Avg_Ticket_Price=("Average_Ticket_Price", "mean"),
          Avg_Satisfaction=("Passenger_Satisfaction", "mean"),
          Avg_Baggage_Kg=("Avg_Baggage_Kg", "mean")
      )
      .sort_values("Passengers", ascending=False)
      .round(2)
)

display(class_summary)
```

```python
weather_summary = (
    df.groupby("Weather")
      .agg(
          Flights=("Flight_ID", "count"),
          Avg_Delay_Min=("Delay_Minutes", "mean"),
          Avg_Satisfaction=("Passenger_Satisfaction", "mean")
      )
      .sort_values("Avg_Delay_Min", ascending=False)
      .round(2)
)

display(weather_summary)
```

```python
route_summary = (
    df.groupby(["Origin", "Destination"])
      .agg(
          Flights=("Flight_ID", "count"),
          Passengers=("Passengers", "sum"),
          Avg_Delay_Min=("Delay_Minutes", "mean"),
          Avg_Satisfaction=("Passenger_Satisfaction", "mean"),
          Avg_Load_Factor=("Load_Factor_%", "mean")
      )
      .sort_values("Passengers", ascending=False)
      .round(2)
)

display(route_summary.head(15))
```

```python
channel_summary = (
    df.groupby("Booking_Channel")
      .agg(
          Flights=("Flight_ID", "count"),
          Avg_Ticket_Price=("Average_Ticket_Price", "mean"),
          Avg_Satisfaction=("Passenger_Satisfaction", "mean"),
          Avg_Delay_Min=("Delay_Minutes", "mean")
      )
      .sort_values("Flights", ascending=False)
      .round(2)
)

display(channel_summary)
```

```python
top_delays = df.nlargest(10, "Delay_Minutes")[
    ["Flight_ID", "Airline", "Origin", "Destination",
     "Delay_Minutes", "Weather", "Flight_Status"]
]

display(top_delays)
```
