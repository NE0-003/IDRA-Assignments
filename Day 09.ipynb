```python
import pandas as pd
import numpy as np

from google.colab import files

uploaded = files.upload()

orders = pd.read_csv("Day9_Orders.csv")
customers = pd.read_csv("Day9_Customers.csv")
products = pd.read_csv("Day9_Products.csv")

print("Orders shape:", orders.shape)
print("Customers shape:", customers.shape)
print("Products shape:", products.shape)

display(orders.head())
display(customers.head())
display(products.head())
```

```python
print("Orders columns:")
print(orders.columns.tolist())
print("\nCustomers columns:")
print(customers.columns.tolist())
print("\nProducts columns:")
print(products.columns.tolist())
```

```python
print("Orders info:")
orders.info()

print("\nMissing values in Orders:")
display(orders.isna().sum().to_frame("Missing_Count"))

print("\nMissing values in Customers:")
display(customers.isna().sum().to_frame("Missing_Count"))

print("\nMissing values in Products:")
display(products.isna().sum().to_frame("Missing_Count"))
```

```python
orders["Order_Date"] = pd.to_datetime(orders["Order_Date"], errors="coerce")

print("Order_Date data type:", orders["Order_Date"].dtype)
print("Minimum order date:", orders["Order_Date"].min())
print("Maximum order date:", orders["Order_Date"].max())

display(orders.head())
```

```python
orders_customers = orders.merge(
    customers,
    on="Customer_ID",
    how="left",
    validate="many_to_one"
)

print("Shape after customer merge:", orders_customers.shape)
display(orders_customers.head())
```

```python
processed = orders_customers.merge(
    products,
    on="Product_ID",
    how="left",
    validate="many_to_one"
)

print("Shape after both merges:", processed.shape)
display(processed.head())
```

```python
orders_part1 = orders.iloc[:len(orders)//2].copy()
orders_part2 = orders.iloc[len(orders)//2:].copy()

orders_concat_demo = pd.concat(
    [orders_part1, orders_part2],
    ignore_index=True
)

print("Part 1 shape:", orders_part1.shape)
print("Part 2 shape:", orders_part2.shape)
print("Concatenated shape:", orders_concat_demo.shape)

print("Same number of Order_ID values:",
      len(orders_concat_demo) == len(orders))
```

```python
processed["Customer_Label"] = processed.apply(
    lambda row: f"{row['Customer_Name']} ({row['Membership_Type']})",
    axis=1
)

processed["Order_Value"] = processed.apply(
    lambda row: row["Quantity"] * row["Unit_Price"],
    axis=1
)

processed["Order_Size"] = processed["Quantity"].apply(
    lambda q: "Bulk" if q >= 5 else ("Medium" if q >= 3 else "Small")
)

display(
    processed[
        ["Order_ID", "Customer_Label", "Quantity", "Unit_Price",
         "Order_Value", "Order_Size"]
    ].head(10)
)
```

```python
processed["Order_Year"] = processed["Order_Date"].dt.year
processed["Order_Month"] = processed["Order_Date"].dt.month
processed["Order_Month_Name"] = processed["Order_Date"].dt.month_name()
processed["Order_Day"] = processed["Order_Date"].dt.day
processed["Order_Day_Name"] = processed["Order_Date"].dt.day_name()
processed["Order_Week"] = processed["Order_Date"].dt.isocalendar().week.astype("Int64")

display(
    processed[
        ["Order_ID", "Order_Date", "Order_Year", "Order_Month",
         "Order_Month_Name", "Order_Day", "Order_Day_Name", "Order_Week"]
    ].head(10)
)
```

```python
final_columns = [
    "Order_ID", "Order_Date", "Order_Year", "Order_Month", "Order_Month_Name",
    "Order_Day", "Order_Day_Name", "Order_Week",
    "Customer_ID", "Customer_Name", "Customer_Label", "City", "Region",
    "Membership_Type", "Product_ID", "Product_Name", "Category", "Brand",
    "Unit_Price", "Quantity", "Order_Value", "Order_Size",
    "Payment_Method", "Order_Status"
]

processed = processed[final_columns].sort_values(
    ["Order_Date", "Order_ID"]
).reset_index(drop=True)

display(processed.head(10))
print("Final shape:", processed.shape)
```

```python
print("Missing customer names:", processed["Customer_Name"].isna().sum())
print("Missing product names:", processed["Product_Name"].isna().sum())
print("Duplicate Order_ID values:", processed["Order_ID"].duplicated().sum())

print("\nFinal data types:")
display(processed.dtypes.to_frame("Data_Type"))

print("\nFinal descriptive statistics:")
display(processed[["Quantity", "Unit_Price", "Order_Value"]].describe().round(2))
```

```python
print("Total orders:", processed["Order_ID"].nunique())
print("Total customers:", processed["Customer_ID"].nunique())
print("Total products:", processed["Product_ID"].nunique())
print("Total quantity ordered:", processed["Quantity"].sum())
print("Total order value:", round(processed["Order_Value"].sum(), 2))

print("\nOrders by membership type:")
display(processed["Membership_Type"].value_counts().to_frame("Order_Count"))

print("\nOrders by category:")
display(processed["Category"].value_counts().to_frame("Order_Count"))

print("\nOrder value by month:")
display(
    processed.groupby(["Order_Year", "Order_Month", "Order_Month_Name"])["Order_Value"]
    .sum()
    .sort_index()
    .round(2)
    .to_frame("Total_Order_Value")
)
```

```python
output_file = "Day9_Processed_Ecommerce_Dataset.csv"

processed.to_csv(output_file, index=False)

print(f"Processed dataset exported successfully to: {output_file}")
display(pd.read_csv(output_file).head())

```
